## Number-theoretic functions

### 1 - Math Combination

math.<b>comb</b>(<i>n, k</i>)

<p>Return the number of ways to choose k items from n items without repetition and without order.</p>
<p>Evaluates to <span style="background-color: gray;">n! / (k! * (n - k)!)</span> when <span style="background-color:gray">k <= n</span> and evaluates to zero when k > n.</p>

<h3>Exercise:</h3>

<summary><b>👉 Click here to see the exercise details</b></summary>
<div>
    <p style="margin-bottom: -15px;"><i>n</i> = 5, <i>k</i> = 2</p>
    <p>n! / (k! * (n - k)!) = 5! / (2! * 3!)</p>
</div>
<div>
    <b style="margin-bottom: -15px;">Counting the "!" Factorial</b>
    <p style="margin-bottom: -10px;">Factorial <i>(denoted by the exclamation mark !)</i> is the product of all positive integers from the specified n before the "!", down to the number below of the n.</p>
    <p style="margin-bottom: -15px;"><b>For Instance: </b>5!</p>
    <p style="margin-bottom: -15px;">Factorial 5 means by 5 down to 4, down to 3, down to 2, and down to 1. So, it counts as:</p>
    <p>5! = 5 * 4 * 3 * 2 * 1 = 120</p>
</div>
<b>Let's Count the Equation Above:</b>
<p>
    5! / (2! * 3!)<br>
    = 120 / (2 * 6)<br>
    = 120 / 12<br>
    = 10
</p>
</details>


<p style="margin-bottom: -15px;">Also called the binomial coefficient because it is equivalent to the coefficient of k-th term in polynomial expansion of (1 + x)ⁿ.</p>
<p>Raises TypeError if either of the arguments are not integers. Raises ValueError if either of the arguments are negative.</p>

### 3 - Greatest Common Divisor (GCD)
<p>math.<b>gcd</b>( <i>*integers</i> )</p>

<p>Return the greatest common divisor of the specified integer arguments. If any of the arguments is nonzero, then the returned value is the largest positive integer that is a divisor of all arguments.</p>
<div>
    <p>It is calculated using the <b>Euclidean Algorithm:</b></p>
    <details>
    <summary><b>👉 Click here to see how to count GCD and the examples behind</b></summary>
        <ul>
            <li>For two numbers, repeatedly replace the larger number with the remainder of dividing the larger by the smaller until one becomes zero; the non‑zero number is the GCD.</li>
            <li>For more than two numbers, it computes the GCD progressively:<br><span style="background-color:gray;">&nbsp;gcd(a, b, c) = gcd(gcd(a, b), c) </span>.</li>
        </ul>
        <h4>Examples:</h4>
        <ul>
            <li><span style="background-color:gray;">&nbsp;math.gcd(12, 18) </span> -> 6 <i>(since 6 divides both 12 and 18, and no larger integer does)</i></li>
            <li><span style="background-color:gray;">&nbsp;math.gcd(15, 25, 35) </span> -> 5</li>
            <li><span style="background-color:gray;">&nbsp;math.gcd(0, 7) </span> -> 7 <i>(any number divides 0, so the GCD is the non-zero number)</i></li>
            <li><span style="background-color:gray;">&nbsp;math.gcd(0, 0)) </span> -> 0 <i>(special case when all arguments are zero)</i></li>
        </ul>
    </summary>
</div>

```python
math.gcd(0, 12)
```

<p>If all arguments are zero, then the returned value is 0. gcd() without arguments returns 0.</p>
